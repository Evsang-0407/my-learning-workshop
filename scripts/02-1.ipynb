```python
import os
from dotenv import load_dotenv
load_dotenv()

api_key = os.getenv("ZAI_API_KEY")
if not api_key:
    raise ValueError("未在 .env 文件中找到 ZAI_API_KEY，请检查文件")

print(f"成功读取 API Key: {api_key[:8]}...") # 仅打印前8位用于验证
```

```python
# API key 从 .env 文件读取（见上一个 cell），不要硬编码：
# .env 文件中写一行：ZAI_API_KEY=你的密钥
# 提醒：切勿把真实密钥写进 notebook，尤其是要推送到公开仓库的文件
```

```python
from zai import ZhipuAiClient

# 初始化客户端
client = ZhipuAiClient(api_key=api_key)

# 发起对话
response = client.chat.completions.create(
    model="glm-4", # 当前稳定可用模型
    messages=[
        {"role": "system", "content": "你是一个有用的AI助手。"},
        {"role": "user", "content": "你好，请介绍一下自己。"}
    ],
    temperature=0.6,
)

print(response.choices[0].message.content)
```
